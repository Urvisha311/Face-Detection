```python
face_path = r"/face.jpg"
face_xml = r"/face.xml"
face_cascade = cv2.CascadeClassifier(face_xml)
face_img = cv2.imread(face_path)
face_img = cv2.cvtColor(face_img,cv2.COLOR_BGR2RGB)
faces = face_cascade.detectMultiScale(face_img)
for x,y,w,h in faces:
    cv2.rectangle(face_img,(x,y),(x+w,y+h),(255,0,0),3)
plt.figure(figsize=(6,15))
plt.imshow(face_img)
plt.show()
print(faces)
```

```python
face_path = r"/face.jpg"
face_xml = r"/face.xml"
eye_xml = r"/eye.xml"
face_cascade = cv2.CascadeClassifier(face_xml)
eye_cascade = cv2.CascadeClassifier(eye_xml)
face_img = cv2.imread(face_path)
face_img = cv2.cvtColor(face_img,cv2.COLOR_BGR2RGB)
faces = face_cascade.detectMultiScale(face_img)
for x,y,w,h in faces:
    cv2.rectangle(face_img,(x,y),(x+w,y+h),(255,0,0),3)
    single_face = face_img[y:y+h,x:x+w]
    eyes = eye_cascade.detectMultiScale(single_face,1.2,10)
    for ex,ey,ew,eh in eyes:
        cv2.rectangle(single_face,(ex,ey),(ex+ew,ey+eh),(0,0,255),3)
plt.figure(figsize=(6,15))
plt.imshow(face_img)
plt.show()
print(faces)
```

### Eyes Detection

```python
### Access the Webcam
import cv2
cam = cv2.VideoCapture(0)
face_xml = r"C:\Users\UMANG KEJRIWAL\Desktop\Open CV\face.xml"
eye_xml = r"C:\Users\UMANG KEJRIWAL\Desktop\Open CV\eye.xml"
face_cascade = cv2.CascadeClassifier(face_xml)
eye_cascade = cv2.CascadeClassifier(eye_xml)
while True:
    ret,img = cam.read()
    faces = face_cascade.detectMultiScale(img)
    for x,y,w,h in faces:
        cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),3)
        single_face = img[y:y+h,x:x+w]
        eyes = eye_cascade.detectMultiScale(single_face,1.2)
        for ex,ey,ew,eh in eyes:
            cv2.rectangle(single_face,(ex,ey),(ex+ew,ey+eh),(0,0,255),3)
    cv2.imshow("WebCam",img)
    if cv2.waitKey(1) == 13:          ### ascii code for enter
        break
cam.release()
cv2.destroyAllWindows()
```

```python
### Object Detection by Color (Blue Colour)
import cv2
import numpy as np
cam = cv2.VideoCapture(0)
while True:
    ret,img = cam.read()
    hsv = cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
    #blue
    low = np.array([100,50,50])
    high = np.array([140,255,255])
    mask_image = cv2.inRange(hsv,low,high)
    final_image = cv2.bitwise_and(img,img,mask=mask_image)
    cv2.imshow("Final Image",final_image)
    cv2.imshow("Mask Image",mask_image)
    cv2.imshow("HSV",hsv)  ### hue saturation value
    cv2.imshow("WebCam",img)
    if cv2.waitKey(1) == 13:          ### ascii code for enter
        break
cam.release()
cv2.destroyAllWindows()
```
